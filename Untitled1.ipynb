```python
%%writefile inference.py

import torch
import cv2
import numpy as np
import argparse
import os

parser = argparse.ArgumentParser()
parser.add_argument('--image', type=str, required=True)
parser.add_argument('--checkpoint', type=str, required=True)
args = parser.parse_args()

# Load image
image = cv2.imread(args.image, 0)

# Improve contrast
image = cv2.equalizeHist(image)

# Normalize
image = image / 255.0
image = torch.tensor(image).unsqueeze(0).unsqueeze(0).float()

# Load model
model = torch.load(args.checkpoint, map_location='cpu')
model.eval()

with torch.no_grad():
    prob_map = model(image)[0][0].numpy()

# Threshold
threshold = 0.7
binary_map = prob_map > threshold

# Save outputs
os.makedirs("outputs", exist_ok=True)

cv2.imwrite("outputs/prob_map.png", prob_map * 255)
cv2.imwrite("outputs/detected.png", (binary_map * 255).astype(np.uint8))

print("Saved improved outputs in 'outputs/' folder")
```
